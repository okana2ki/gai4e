```python
css_content = """
body {
    background-color: #2c3e50;
    color: #ecf0f1;
    font-family: 'Helvetica Neue', Arial, sans-serif;
    display: flex;
    justify-content: center;
    align-items: center;
    height: 100vh;
    margin: 0;
    transition: background-color 0.5s ease;
}
.container {
    background-color: #34495e;
    padding: 2rem;
    border-radius: 10px;
    box-shadow: 0 4px 6px rgba(0,0,0,0.3);
    text-align: center;
}
h1 {
    color: #3498db;
}
button {
    background-color: #e67e22;
    color: white;
    border: none;
    padding: 10px 20px;
    border-radius: 5px;
    cursor: pointer;
    font-size: 1rem;
    margin-top: 20px;
}
button:hover {
    background-color: #d35400;
}
"""

js_content = """
function changeColor() {
    const colors = ['#2c3e50', '#8e44ad', '#27ae60', '#d35400', '#c0392b'];
    const randomColor = colors[Math.floor(Math.random() * colors.length)];
    document.body.style.backgroundColor = randomColor;
}
"""

html_content = """
<!DOCTYPE html>
<html lang='ja'>
<head>
    <meta charset='UTF-8'>
    <title>Interactive Web Page</title>
    <link rel='stylesheet' href='style.css'>
</head>
<body>
    <div class='container'>
        <h1>Interactive Page</h1>
        <p>JavaScriptを使って動きをつけました。</p>
        <button onclick='changeColor()'>背景色を変える</button>
    </div>
    <script src='script.js'></script>
</body>
</html>
"""

with open('style.css', 'w') as f:
    f.write(css_content)

with open('script.js', 'w') as f:
    f.write(js_content)

with open('index.html', 'w') as f:
    f.write(html_content)

print('index.html, style.css, script.js を作成しました。')
```

### 改善案：モダンな構成（リファクタリング版）

以下の改善を行いました：
- **CSS変数 (--primary-color等)** を導入し、デザインの統一管理を可能にしました。
- **JSの分離**: HTMLに `onclick` を書かず、JavaScript側でイベントを監視するようにしました。
- **レスポンシブ対応**: モバイル端末でも見やすいよう調整しています。

```python
css_modern = """
:root {
    --bg-color: #2c3e50;
    --card-bg: #34495e;
    --text-color: #ecf0f1;
    --accent-color: #3498db;
    --button-color: #e67e22;
    --button-hover: #d35400;
}

body {
    background-color: var(--bg-color);
    color: var(--text-color);
    font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
    display: grid;
    place-items: center;
    min-height: 100vh;
    margin: 0;
    transition: background-color 0.6s cubic-bezier(0.4, 0, 0.2, 1);
}

.container {
    background-color: var(--card-bg);
    padding: 3rem;
    border-radius: 16px;
    box-shadow: 0 10px 30px rgba(0,0,0,0.4);
    text-align: center;
    max-width: 90%;
    width: 400px;
}

h1 {
    color: var(--accent-color);
    margin-bottom: 1.5rem;
}

button {
    background-color: var(--button-color);
    color: white;
    border: none;
    padding: 12px 24px;
    border-radius: 8px;
    cursor: pointer;
    font-size: 1.1rem;
    font-weight: bold;
    transition: transform 0.2s, background-color 0.2s;
}

button:hover {
    background-color: var(--button-hover);
    transform: translateY(-2px);
}

button:active {
    transform: translateY(0);
}
"""

js_modern = """
document.addEventListener('DOMContentLoaded', () => {
    const btn = document.querySelector('#colorBtn');
    const colors = ['#2c3e50', '#8e44ad', '#27ae60', '#d35400', '#c0392b', '#16a085', '#2980b9'];

    btn.addEventListener('click', () => {
        const randomColor = colors[Math.floor(Math.random() * colors.length)];
        document.body.style.backgroundColor = randomColor;
    });
});
"""

html_modern = """
<!DOCTYPE html>
<html lang='ja'>
<head>
    <meta charset='UTF-8'>
    <meta name='viewport' content='width=device-width, initial-scale=1.0'>
    <title>Modern Interactive Page</title>
    <link rel='stylesheet' href='style.css'>
</head>
<body>
    <main class='container'>
        <h1>Modern Page</h1>
        <p>ベストプラクティスを適用した構成です。</p>
        <button id='colorBtn'>背景色をランダムに変更</button>
    </main>
    <script src='script.js'></script>
</body>
</html>
"""

with open('style.css', 'w') as f: f.write(css_modern)
with open('script.js', 'w') as f: f.write(js_modern)
with open('index.html', 'w') as f: f.write(html_modern)

print('モダン版の index.html, style.css, script.js を上書き作成しました。')
```
